```python

import numpy as np
import matplotlib.pyplot as plt


np.random.seed(42)

def generate_fruit_dataset(n_samples=300):
    samples_per_class = n_samples // 2


    weight_apples = np.random.normal(loc=150, scale=15, size=samples_per_class)
    diameter_apples = np.random.normal(loc=7.0, scale=0.5, size=samples_per_class)
    redness_apples = np.random.normal(loc=0.85, scale=0.08, size=samples_per_class)
    class_apples = np.zeros(samples_per_class)


    weight_oranges = np.random.normal(loc=200, scale=20, size=samples_per_class)
    diameter_oranges = np.random.normal(loc=8.5, scale=0.6, size=samples_per_class)
    redness_oranges = np.random.normal(loc=0.25, scale=0.10, size=samples_per_class)
    class_oranges = np.ones(samples_per_class)


    X_raw = np.vstack((
        np.column_stack((weight_apples, diameter_apples, redness_apples)),
        np.column_stack((weight_oranges, diameter_oranges, redness_oranges))
    ))
    y = np.hstack((class_apples, class_oranges))

    mean = np.mean(X_raw, axis=0)
    std = np.std(X_raw, axis=0)
    X_scaled = (X_raw - mean) / std

    return X_raw, X_scaled, y


X_raw, X, y = generate_fruit_dataset(n_samples=300)
print(f"Dataset generated: X shape = {X.shape}, y shape = {y.shape}")
print(f"Features: [Weight (scaled), Diameter (scaled), Color Index (scaled)]")
```

```python
class SingleNeuronNetwork:
    def __init__(self, input_dim=3, seed=42):
        np.random.seed(seed)
        self.w = np.random.randn(input_dim)
        self.b = np.random.randn()

    def sigmoid(self, z):
        z = np.clip(z, -500, 500)
        return 1.0 / (1.0 + np.exp(-z))

    def forward(self, X):
        z = np.dot(X, self.w) + self.b
        return self.sigmoid(z)

    def compute_loss(self, y_true, y_pred):
        eps = 1e-15
        y_pred = np.clip(y_pred, eps, 1 - eps)
        loss = -np.mean(y_true * np.log(y_pred) + (1 - y_true) * np.log(1 - y_pred))
        return loss

    def compute_accuracy(self, y_true, y_pred):
        predictions = (y_pred >= 0.5).astype(int)
        return np.mean(predictions == y_true)

    def train_gradient_descent(self, X, y, lr=0.1, epochs=200):
        history = {'loss': [], 'accuracy': []}
        N = X.shape[0]

        for epoch in range(epochs):
            y_pred = self.forward(X)

            loss = self.compute_loss(y, y_pred)
            acc = self.compute_accuracy(y, y_pred)
            history['loss'].append(loss)
            history['accuracy'].append(acc)

            dz = y_pred - y
            dw = (1 / N) * np.dot(X.T, dz)
            db = (1 / N) * np.sum(dz)

            self.w -= lr * dw
            self.b -= lr * db

        return history

    def train_random_search(self, X, y, epochs=200):
        history = {'loss': [], 'accuracy': []}
        best_loss = float('inf')

        for epoch in range(epochs):
            w_candidate = np.random.randn(X.shape[1])
            b_candidate = np.random.randn()

            z = np.dot(X, w_candidate) + b_candidate
            y_pred = self.sigmoid(z)

            loss = self.compute_loss(y, y_pred)
            acc = self.compute_accuracy(y, y_pred)

            if loss < best_loss:
                best_loss = loss
                self.w = w_candidate
                self.b = b_candidate

            history['loss'].append(best_loss)
            history['accuracy'].append(acc)

        return history
```

```python
model_gd = SingleNeuronNetwork(input_dim=3, seed=42)
initial_w_gd = model_gd.w.copy()
initial_b_gd = model_gd.b

print("--- BEFORE TRAINING ---")
print(f"Initial Weights (GD): {initial_w_gd}")
print(f"Initial Bias (GD):    {initial_b_gd:.4f}\n")

history_gd = model_gd.train_gradient_descent(X, y, lr=0.1, epochs=200)

print("--- AFTER TRAINING (Gradient Descent) ---")
print(f"Final Weights (GD):   {model_gd.w}")
print(f"Final Bias (GD):      {model_gd.b:.4f}")
print(f"Final Accuracy:       {history_gd['accuracy'][-1] * 100:.2f}%\n")

model_no_gd = SingleNeuronNetwork(input_dim=3, seed=42)
history_no_gd = model_no_gd.train_random_search(X, y, epochs=200)

print("--- AFTER TRAINING (Without Gradient Descent) ---")
print(f"Final Accuracy (No GD): {history_no_gd['accuracy'][-1] * 100:.2f}%")
```

```python
def compute_confusion_matrix(y_true, y_pred_prob):
    y_pred = (y_pred_prob >= 0.5).astype(int)
    tp = np.sum((y_true == 1) & (y_pred == 1))
    tn = np.sum((y_true == 0) & (y_pred == 0))
    fp = np.sum((y_true == 0) & (y_pred == 1))
    fn = np.sum((y_true == 1) & (y_pred == 0))
    return np.array([[tn, fp], [fn, tp]])

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(history_gd['accuracy'], label='With Gradient Descent', color='green', linewidth=2)
axes[0].plot(history_no_gd['accuracy'], label='Without Gradient Descent', color='red', linestyle='--')
axes[0].set_title('Epoch vs. Model Accuracy', fontsize=12, fontweight='bold')
axes[0].set_xlabel('Epochs')
axes[0].set_ylabel('Accuracy Score')
axes[0].grid(True, alpha=0.3)
axes[0].legend()

axes[1].plot(history_gd['loss'], label='With Gradient Descent', color='blue', linewidth=2)
axes[1].plot(history_no_gd['loss'], label='Without Gradient Descent', color='orange', linestyle='--')
axes[1].set_title('Epoch vs. Binary Cross-Entropy Loss', fontsize=12, fontweight='bold')
axes[1].set_xlabel('Epochs')
axes[1].set_ylabel('Loss')
axes[1].grid(True, alpha=0.3)
axes[1].legend()

plt.tight_layout()
plt.show()

cm_gd = compute_confusion_matrix(y, model_gd.forward(X))

fig, ax = plt.subplots(figsize=(5, 4))
im = ax.imshow(cm_gd, cmap='Blues', interpolation='nearest')
plt.colorbar(im)
ax.set_xticks([0, 1])
ax.set_yticks([0, 1])
ax.set_xticklabels(['Apple (0)', 'Orange (1)'])
ax.set_yticklabels(['Apple (0)', 'Orange (1)'])
ax.set_xlabel('Predicted Label', fontweight='bold')
ax.set_ylabel('True Label', fontweight='bold')
ax.set_title('Confusion Matrix (Gradient Descent)', fontweight='bold')

for i in range(2):
    for j in range(2):
        ax.text(j, i, str(cm_gd[i, j]), ha='center', va='center', color='black', fontsize=14, fontweight='bold')

plt.tight_layout()
plt.show()
```

```python
def plot_3d_decision_space(X, y, weights, bias, title):
    fig = plt.figure(figsize=(9, 7))
    ax = fig.add_subplot(111, projection='3d')

    apples = X[y == 0]
    oranges = X[y == 1]
    ax.scatter(apples[:, 0], apples[:, 1], apples[:, 2], color='red', label='Apples (Class 0)', alpha=0.6, edgecolors='k')
    ax.scatter(oranges[:, 0], oranges[:, 1], oranges[:, 2], color='orange', label='Oranges (Class 1)', alpha=0.6, edgecolors='k')

    x1_span = np.linspace(X[:, 0].min(), X[:, 0].max(), 12)
    x2_span = np.linspace(X[:, 1].min(), X[:, 1].max(), 12)
    X1_mesh, X2_mesh = np.meshgrid(x1_span, x2_span)

    if abs(weights[2]) > 1e-5:
        X3_mesh = (-weights[0] * X1_mesh - weights[1] * X2_mesh - bias) / weights[2]
        ax.plot_surface(X1_mesh, X2_mesh, X3_mesh, alpha=0.3, cmap='viridis')

    ax.quiver(0, 0, 0, weights[0], weights[1], weights[2], color='black', length=1.5, linewidth=3, arrow_length_ratio=0.2, label='Weight Vector (w)')

    ax.set_xlabel('Weight (Scaled)')
    ax.set_ylabel('Diameter (Scaled)')
    ax.set_zlabel('Color Index (Scaled)')
    ax.set_title(f"{title}\nWeights: [{weights[0]:.2f}, {weights[1]:.2f}, {weights[2]:.2f}] | Bias: {bias:.2f}", fontsize=11)
    ax.legend()
    plt.tight_layout()
    plt.show()

plot_3d_decision_space(X, y, initial_w_gd, initial_b_gd, "Feature Space & Decision Boundary BEFORE Gradient Descent")

plot_3d_decision_space(X, y, model_gd.w, model_gd.b, "Feature Space & Decision Boundary AFTER Gradient Descent")
```

```python
learning_rates = [0.001, 0.01, 0.1, 0.5, 1.0]
epoch_counts = [50, 200, 500]

print("=== HYPERPARAMETER EXPLORATION RESULTS ===")
print(f"{'Learning Rate':<15} | {'Epochs':<10} | {'Final Loss':<12} | {'Final Accuracy (%)':<18}")
print("-" * 65)

for lr in learning_rates:
    for ep in epoch_counts:
        test_model = SingleNeuronNetwork(input_dim=3, seed=42)
        hist = test_model.train_gradient_descent(X, y, lr=lr, epochs=ep)
        final_loss = hist['loss'][-1]
        final_acc = hist['accuracy'][-1] * 100
        print(f"{lr:<15} | {ep:<10} | {final_loss:<12.4f} | {final_acc:<18.2f}")
```
